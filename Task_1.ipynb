```python
import torch
import  torch.nn as nn

class TelemetryClassifier(nn.Module):
    def __init__( self , input_features=6):
      super(TelemetryClassifier , self).__init__()
      self.stuff = nn.Sequential(
          nn.Linear(input_features ,16),
          nn.ReLU() ,
          nn.Linear(16 , 8),
          nn.ReLU() ,
          nn.Linear(8 , 1) ,
          nn.Sigmoid()
      )
    def forward(self,x):
      return self.stuff(x)
```

```python
batch_size = 8 ;
num_features = 6 ;

dummy_input = torch.rand(batch_size, num_features)

print("Input Shape:", dummy_input.shape)
print("Sample Data (First 8 users):\n", dummy_input[:8])
model = TelemetryClassifier(input_features=num_features)
model.eval()
with torch.no_grad():
    probabilities = model(dummy_input)

print("\nOutput Shape:", probabilities.shape)
print("Generated Probabilities:\n", probabilities)
```
